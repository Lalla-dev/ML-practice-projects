```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/train.txt',header = None , names = ['text','emotions'], sep = ';')
```

```python
df['emotions'].unique()
```

```python
emotions = df['emotions'].unique()
i = 0
for emo in emotions:
  df['emotions'] = df['emotions'].replace(emo,i)
  i +=1

df
```

```python
df['text'] = df['text'].apply(lambda x : x.lower())
```

```python
import string

def remove_punc(txt):
  return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text'].apply(remove_punc)
```

```python
def remove_txt(txt):
  new = ""
  for i in txt:
    if not i.isdigit():
      new += i
  return new
df['text'] = df['text'].apply(remove_txt)

```

```python
def remove_emoji(emo):
  new =""
  for i in emo:
    if i.isascii():
      new += i
  return new
df['text'] = df['text'].apply(remove_emoji)
df
```

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('stopwords')
nltk.download('punkt')
```

```python
stop_words = set(stopwords.words('english'))
stop_words
```

```python
len(stop_words)
nltk.download('punkt_tab')
```

```python
def remove_sw(txt):
  words = txt.split()
  cleaned=[]
  for i in words:
    if i not in stop_words:
      cleaned.append(i)
  return " ".join(cleaned)
```

```python
df['text'] = df['text'].apply(remove_sw)
```
