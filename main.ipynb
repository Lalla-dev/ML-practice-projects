```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split 
from sklearn.linear_model import LinearRegression 
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('diabetes.csv') 
df.head()
```

```python
x = df['Glucose'] 
y = df['Outcome']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.33, random_state=42)
```

```python
model = LinearRegression()
```

```python
model = model.fit(x_train.values.reshape(-1, 1), y_train)
```

```python
y_pred_continuous = model.predict(x_test.values.reshape(-1, 1)) 
y_pred = (y_pred_continuous >= 0.5).astype(int)
```

```python
accuracy = accuracy_score(y_test, y_pred) 
print(f"Accuracy: {accuracy}")
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(x_test, y_test, color='blue', alpha=0.5, label='Actual Data')
plt.plot(x_test, y_pred_continuous, color='red', linewidth=2, label='Regression Line')

plt.title('Linear Regression: Glucose vs Outcome')
plt.xlabel('Glucose')
plt.ylabel('Outcome')
plt.ylim(-0.1, 1.1)
plt.legend()
plt.grid(True)

plt.show()
```
